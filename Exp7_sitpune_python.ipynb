```python
# Question 7
num = int(input("Enter a number: "))
rev = 0
while num > 0:
  digit = num % 10
  rev = rev * 10 + digit
  num //= 10

print("Reversed Number =", rev)
```

```python
# Question 8
# Check Palidrome Number/String

num = int(input("Enter a Number: "))
temp = num
rev = 0

while num > 0:
    rev = rev * 10 + num % 10
    num = num // 10

if temp == rev:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")




```

```python
# Question 9
# Program to check if a string is a palindrome

s = input("Enter a String: ")
i, j = 0, len(s) - 1
is_palindrome = True

while i < j:
    if s[i] != s[j]:
        is_palindrome = False
        break
    i += 1
    j -= 1

if is_palindrome:
    print("Palindrome String")
else:
    print("Not a Palindrome String")
```

```python
# Program to check if a string is a palindrome using slicing
st = input("Enter the String: ")
rev = st[::-1]

if st == rev:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
# Question 10
# Count digits in a number using while loop

num = int(input("Enter a Number: "))
count = 0

while num > 0:
  count += 1
  num //= 10

print("Number of digits =", count)

```

***The break statement***

With the break statement we can stop the loop even if the while condition is true

```python
#code to exit the loop when i is 3

i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
```

```python
# Program to search for an element in a list
nums = [10, 20, 30, 40, 50]
key = int(input("Enter the element to search: "))
i = 0

while i < len(nums):
    if nums[i] == key:
        print("element found at index", i + 1 )
        break
    i += 1
else:
    print("element not found")
```

```python
# continue statement (skips the loop)
# print only odd numbers (1 to 10)

i = 0
while i < 10:
  i += 1
  if i % 2 == 0:
    continue
  print(i)
```
